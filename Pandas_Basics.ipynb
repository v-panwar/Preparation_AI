# Step 4: Data Cleaning and Manipulation

```python
df=pd.read_csv('sample_data.csv')
print(df)
```

```python
df_null = df.isnull()
print("Rows with null values:\n", df_null)
print("Total Count with null values:\n", df_null.sum().sum())

print("Total Count with null values:\n", df_null.sum(axis=0))
```

```python
mean_age = df['AGE'].mean()
df_filled = df.copy()
df_filled['AGE']  = df_filled['AGE'].fillna(mean_age)
df_filled['AGE_in_Months']  = (df_filled['AGE']*12).astype('int64')  # Convert age to months
df_filled['SCORE']= df_filled['SCORE'].fillna(0)
df_filled['NAME'] = df_filled['NAME'].str.upper()
# print("Dataframe with SCORE filled with static value:\n", df_filled_age['SCORE'].fillna(0, inplace=True))
conditions = [
    df_filled['CITY'].isin(['New York', 'Philadelphia', 'Chicago']),
    df_filled['CITY'].isin(['Los Angeles', 'San Francisco', 'San Diego', 'San Jose']),
    df_filled['CITY'].isin(['Houston', 'Dallas', 'Austin', 'San Antonio', 'Phoenix']),
]
choices = ['East coast', 'West coast', 'South West']
df_filled['REGION'] = np.select(conditions, choices, default='Other')
avg_by_region = df_filled.groupby('REGION').agg({'SCORE': 'mean', 'AGE': 'mean', 'ID': 'count'}).rename(columns={'ID': 'Count', 'SCORE': 'Mean_score', 'AGE': 'Mean_Age'})

print(f"\nDataFrame after filling Names with Upper NAMEs, missing score 0 for NaN and Age with mean ({mean_age:.2f}):\n", df_filled)
print("Average SCORE, AGE and count of students by REGION:\n", avg_by_region)
```
