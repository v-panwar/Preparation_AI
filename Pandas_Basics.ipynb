```python
import pandas as pd
import numpy as np

print("----1.1:-------Creating a pandas series-----------")
s=pd.Series([10, 20, 30, 40, 50])  
print("My series :\n", s)
print(("type of s: ", type(s)))
print("Series index: ", s.index)
print("Servies values: ", s.values)
```

```python
# Series with custom index
s_labeled = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])
print("Series with custom index:\n", s_labeled)
print("Value at index 'c':", s_labeled['c'])
```

```python
# Pandas data frame creation
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [25, 30, 35, 40],
        'City':['New York', 'SFO', 'Los Angeles', 'Chicago'],
        'Salary': [70000, 80000, 120000, 90000]
        }
df = pd.DataFrame(data)
print("Dataframe from dictionary:\n", df)
print("shape of the dataframe: ", df.shape)
print("columns of the dataframe: ", df.columns)
print("indexes of the dataframe: ", df.index)
print("Dataframe values:\n", df.values)
print("First 2 rows of the dataframe:\n", df.head(2))
print("Data types of the dataframe:\n", df.dtypes)  
```
